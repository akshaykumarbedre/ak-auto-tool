```python
from llama_index.core import VectorStoreIndex, DocumentSummaryIndex

# Initialize empty indices
doc_sum_index = DocumentSummaryIndex.from_documents([])
vector_index = VectorStoreIndex.from_documents([]) 

# Now add documents to the indices 
for doc in documents:
    doc_sum_index.insert(doc)
    vector_index.insert(doc)
```

```python
summary_query_engine=doc_sum_index.as_query_engine()
docu_query_engine=vector_index.as_query_engine()
```

```python

print(summary_query_engine.query("summary the reference   ").response)
```

```python
from llama_index.core.tools import QueryEngineTool, ToolMetadata

tools1 = QueryEngineTool(
    query_engine=docu_query_engine, 
    metadata=ToolMetadata(
        name="Retriver_tool",
        description="Retrieves detailed information from the complete PDF based on your query. Use this tool when you need specific facts, details, or in-depth information from the document.",
    ),
)

tools2 = QueryEngineTool(
    query_engine=summary_query_engine, 
    metadata=ToolMetadata(
        name="Summarizer_tool",
        description="Provides concise summaries on specific topics mentioned in the document. Use this tool when you need a high-level overview or summary of a particular subject or section.",
    ),
)
```

```python
tools=[tools2 ,tools1]
```

```python
from llama_index.core.agent.workflow import FunctionAgent
agent = FunctionAgent(tools=tools,llm=llm,system_prompt="You are a helpful assistant use the tools to get the answer based your query.")
```

```python
user_quer="tell me about architecture"
response = await agent.run(user_quer)
print(str(response))
```

```python
from langchain_community.utilities import ArxivAPIWrapper
def fetch_research_paper(paper_id_or_name: str):
    """
    Fetches a research paper from arXiv using either the paper ID or title.
    
    Args:
        paper_id_or_name (str): The paper ID (e.g., "1605.08386") or paper title
        
    Returns:
        The fetched paper document(s)
    """
    arxiv = ArxivAPIWrapper(
    top_k_results = 30,
    ARXIV_MAX_QUERY_LENGTH = 30000,
    load_max_docs = 2000,
    doc_content_chars_max = 40000
)

    docs = arxiv.run(paper_id_or_name)
    return docs

# Example usage
paper = fetch_research_paper(" is all you need")
paper

```
